# Tool Calling · Notebook 05 — Expected vs LLM vs T5, One Query at a Time

Pick a query, and see three answers side by side:

| | Where it comes from |
|---|---|
| **Expected** | The gold set (Notebook 02), written by people before any model ran |
| **LLM** | `gpt-5.6-luna`, given all 18 tool definitions |
| **T5** | Our fine-tuned FLAN-T5-small (Notebook 04), on the CPU |

> **The LLM doesn't *run* any tool.** It only **chooses** which tools to call and fills in their arguments. In a real assistant our code would then run them (look up the bill, activate the pack). Here we just look at the choice, because the choice is what we want to check.

|---|---|
| 1 | Setup |
| 2 | Asking the LLM: send the query with the 18 tools |
| 3 | Asking T5 |

## 1 · Setup

```python
# Run once if needed:
# %pip install -q torch transformers sentencepiece pandas openai python-dotenv ipywidgets

import json
import time
from pathlib import Path

import ipywidgets as widgets
import pandas as pd
import torch
from dotenv import load_dotenv
from IPython.display import display
from openai import OpenAI
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

TOPIC_DIR = Path.cwd().parents[2]
TOOLS_DIR = TOPIC_DIR / "data" / "tools"
T5_DIR    = Path.cwd().parents[1] / "models" / "flan-t5-small-tools"     # saved by Notebook 04

LLM_MODEL = "gpt-5.6-luna"
REASONING = "none"           # reasoning off. If the model rejects "none", use "minimal" or "low"
PREFIX    = "tools: "        # the same prefix T5 was trained with

load_dotenv()                # reads OPENAI_API_KEY from a .env file
client = OpenAI()

catalogue = json.load(open(TOOLS_DIR / "tools.json", encoding="utf-8"))
TOOL_ORDER = catalogue["tool_order"]
OPENAI_TOOLS = json.load(open(TOOLS_DIR / "tools_openai.json", encoding="utf-8"))

gold = pd.read_csv(TOOLS_DIR / "gold_tool_test.csv", keep_default_na=False)
print(len(OPENAI_TOOLS), "tools ·", len(gold), "gold test queries")
```

The same helper as Notebook 04: keep only real tool names, in the fixed catalogue order.

```python
def clean_tools(text):
    parts = []
    for part in text.split(","):
        parts.append(part.strip().lower())

    ordered = []
    for name in TOOL_ORDER:
        if name in parts:
            ordered.append(name)

    if len(ordered) > 0:
        return ", ".join(ordered)
    if "none" in parts:
        return "none"
    return ""

print(clean_tools("get_bill, search_help_articles"))
```

## 2 · Asking the LLM: send the query with the 18 tools

We send three things:
- **instructions**: a short description of the assistant's job,
- **the customer's message**,
- **the 18 tool definitions** from `tools_openai.json` (name, description, parameters).

The LLM's answer is a list of **output items**. Each tool it wants to call comes back as a `function_call` item with a `name` and its `arguments` (as JSON text). If it decides no tool is needed, it just writes a reply.

```python
INSTRUCTIONS = '''You are Grokkers Assist, the customer support assistant for Grokkers Mobile, an Indian telecom operator.
Call every tool the customer's message needs. Call several tools when several are needed.
If the message is unclear, call ask_clarifying_question.
If no tool is needed (a greeting, thanks, or a request outside Grokkers support or against the rules), reply briefly without calling any tool.
For changes to the account, call the tool: the app asks the customer to confirm before anything runs.'''


def ask_llm(query):
    start = time.perf_counter()
    response = client.responses.create(
        model=LLM_MODEL,
        instructions=INSTRUCTIONS,
        input=query,
        tools=OPENAI_TOOLS,
        reasoning={"effort": REASONING},
    )
    seconds = time.perf_counter() - start

    names = []
    calls = []
    for item in response.output:
        if item.type == "function_call":
            names.append(item.name)
            calls.append(item.name + "(" + item.arguments + ")")

    if len(names) == 0:
        tools = "none"
    else:
        tools = clean_tools(", ".join(names))

    return {"query": query, "tools": tools, "calls": " · ".join(calls),
            "reply": response.output_text, "seconds": round(seconds, 2)}
```

**Saving answers.** Each LLM answer is saved to `llm_tool_calls.csv`. Picking the same query again reads the saved answer instead of calling (and paying for) the API twice. Delete the file to ask again.

```python
LLM_FILE = TOOLS_DIR / "llm_tool_calls.csv"

saved_answers = {}
if LLM_FILE.exists():
    for row in pd.read_csv(LLM_FILE, keep_default_na=False).to_dict("records"):
        saved_answers[row["query"]] = row
print(len(saved_answers), "saved LLM answers")


def ask_llm_saved(query):
    if query in saved_answers:
        answer = dict(saved_answers[query])
        answer["seconds_note"] = "saved earlier"
        return answer

    answer = ask_llm(query)
    saved_answers[query] = answer
    pd.DataFrame(list(saved_answers.values())).to_csv(LLM_FILE, index=False)
    answer = dict(answer)
    answer["seconds_note"] = "just now"
    return answer
```

## 3 · Asking T5

```python
tokenizer = AutoTokenizer.from_pretrained(T5_DIR)
t5_model = AutoModelForSeq2SeqLM.from_pretrained(T5_DIR)
t5_model.eval()

def ask_t5(query):
    start = time.perf_counter()
    inputs = tokenizer(PREFIX + query, return_tensors="pt", truncation=True, max_length=96)
    with torch.no_grad():
        ids = t5_model.generate(**inputs, max_new_tokens=48, num_beams=1)
    text = tokenizer.decode(ids[0], skip_special_tokens=True)
    seconds = time.perf_counter() - start
    return {"tools": clean_tools(text), "raw": text, "seconds": round(seconds, 3)}

print(ask_t5("Why is my October bill so high?"))
```

## 4 · Scoring one answer

The rule from Notebook 02: correct if the tools equal the expected `tools` **or** any `accepted` alternative.

```python
def is_correct(predicted, tools, accepted):
    if predicted == tools:
        return True
    if accepted != "":
        for option in accepted.split(" | "):
            if predicted == option:
                return True
    return False

def mark(predicted, tools, accepted):
    if tools == "":
        return ""                       # our own query: no expected answer
    if is_correct(predicted, tools, accepted):
        return "✓"
    return "✗"

print(mark("search_help_articles, get_usage", "search_help_articles, get_usage", ""))
```

## 5 · One example, step by step

Before the widget, one query in plain prints, so every step is visible.

```python
row = gold[gold["query"] == "Why was I charged roaming fees?"].iloc[0]

print("Query:    ", row["query"])
print("Expected: ", row["tools"])
print("Also OK:  ", row["accepted"] or "-")
print()

llm = ask_llm_saved(row["query"])
print("LLM tools:", llm["tools"], mark(llm["tools"], row["tools"], row["accepted"]))
print("LLM calls:", llm["calls"] or "-")
print("LLM reply:", llm["reply"] or "-")
print("LLM time: ", llm["seconds"], "s")
print()

t5 = ask_t5(row["query"])
print("T5 tools: ", t5["tools"], mark(t5["tools"], row["tools"], row["accepted"]))
print("T5 time:  ", t5["seconds"], "s")
```

## 6 · The explorer

- Pick a query from the **gold test set** (never used for training), **or** type your own message in the box. (Your own message has no expected answer, so there's no ✓/✗.)
- Press **Run**.

The panel shows the expected tools and the two models' answers side by side.

```python
def option_label(i, row):
    tag = "original" if row["original"] else row["source"]
    return f"{i + 1:3}. [{tag}] {row['query'][:80]}"

options = []
for i, row in gold.iterrows():
    options.append((option_label(i, row), i))

picker = widgets.Dropdown(options=options, value=0, description="Query:", layout=widgets.Layout(width="95%"))
own_text = widgets.Text(placeholder="...or type your own message here (leave empty to use the dropdown)",
                        layout=widgets.Layout(width="95%"))
run_button = widgets.Button(description="Run", button_style="primary")
panel = widgets.HTML()
```

```python
def tools_html(tools):
    if tools == "":
        return "<i>(nothing valid)</i>"
    if tools == "none":
        return "<i>none</i>"
    parts = []
    for name in tools.split(", "):
        parts.append(f"<code>{name}</code>")
    return " + ".join(parts)

def result_row(who, tools, tick, seconds, extra):
    colour = {"✓": "#2e7d32", "✗": "#c62828"}.get(tick, "#555")
    return (f"<tr><td style='padding:6px 10px'><b>{who}</b></td>"
            f"<td style='padding:6px 10px'>{tools_html(tools)}</td>"
            f"<td style='padding:6px 10px;color:{colour};font-size:18px'><b>{tick}</b></td>"
            f"<td style='padding:6px 10px'>{seconds}</td>"
            f"<td style='padding:6px 10px;font-size:12px;color:#555'>{extra}</td></tr>")

def show(query, expected, accepted, note):
    panel.value = "<i>Asking the LLM and T5…</i>"
    llm = ask_llm_saved(query)
    t5 = ask_t5(query)

    rows = []
    if expected != "":
        also = accepted.replace(" | ", "<br>") if accepted else "-"
        rows.append(f"<tr style='background:#f3f3f3'><td style='padding:6px 10px'><b>Expected</b></td>"
                    f"<td style='padding:6px 10px'>{tools_html(expected)}</td><td></td><td></td>"
                    f"<td style='padding:6px 10px;font-size:12px;color:#555'>also accepted:<br>{also}</td></tr>")

    llm_extra = llm["calls"] if llm["calls"] else "reply: " + llm["reply"][:150]
    rows.append(result_row("LLM", llm["tools"], mark(llm["tools"], expected, accepted),
                           f"{llm['seconds']} s ({llm['seconds_note']})", llm_extra))
    rows.append(result_row("T5", t5["tools"], mark(t5["tools"], expected, accepted),
                           f"{t5['seconds'] * 1000:.0f} ms", "raw output: " + t5["raw"]))

    if llm["tools"] == t5["tools"]:
        verdict = "LLM and T5 <b>agree</b>."
    else:
        verdict = "LLM and T5 <b>disagree</b>. In production, this is the case worth flagging."

    note_html = f"<p style='color:#555'>Note: {note}</p>" if note else ""
    panel.value = (f"<p style='font-size:16px'><b>Customer:</b> {query}</p>"
                   f"<table style='border-collapse:collapse'>"
                   f"<tr><th></th><th style='text-align:left;padding:6px 10px'>Tools</th><th></th>"
                   f"<th style='text-align:left;padding:6px 10px'>Time</th>"
                   f"<th style='text-align:left;padding:6px 10px'>Details</th></tr>"
                   + "".join(rows) + "</table>"
                   f"<p>{verdict}</p>{note_html}")

def on_run(button):
    typed = own_text.value.strip()
    if typed != "":
        show(typed, "", "", "")
    else:
        row = gold.loc[picker.value]
        show(row["query"], row["tools"], row["accepted"], row["note"])

run_button.on_click(on_run)
display(widgets.VBox([picker, own_text, run_button, panel]))
```

**Things to try:**
- **A confusable pair:** "How much data have I used this week?" (`get_usage`) vs "How much data is left in my pack?" (`get_active_packs`).
- **A "how do I" near miss:** "How do I download my postpaid bill as a PDF?" should be an article, not `get_bill`.
- **Two requests in one:** "Change my bill email to [email address] and also tell me when my bill is due."
- **A catalogue gap:** "Activate Wi-Fi calling for me." No tool can do it. Does either model invent or force one?
- **The weak spots from Notebook 04:** catalogue questions ("How many postpaid plans are there in total?"), hand-offs ("Please have a senior executive call me.") and vague messages ("It isn't showing up.").
